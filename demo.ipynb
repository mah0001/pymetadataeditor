# How to use Metadata Editor

```python
from pymetadataeditor import MetadataEditor
import os
```

## Examples of the interface

```python
your_api_key = os.getenv("API_KEY")
api_url = os.getenv("API_URL")
me = MetadataEditor(api_url=api_url, api_key=your_api_key, verify_ssl=False)
```

### Listing your projects

```python
me.list_projects(limit=5)
```

### Creating a new timeseries project

```python
demo_name = "GB20240816_demo"
```

```python
series_description = {
                        "idno": demo_name,
                        "doi": "V1",
                        "name": "Version 1",
                        "display_name": "Version 1"
                     }

timeseries_id = me.create_project_log("timeseries", {"idno": demo_name, "series_description": series_description})
```

```python
me.list_projects(limit=5, sort_by="updated_desc")
```

```python
me.get_project_by_id(timeseries_id)
```

```python
me.get_project_metadata_by_id(timeseries_id, mode='dict')
```

# Starting with outlines

The metadata can be both large and hierarchical. Starting with a skeleton outline makes things easier.

Outlines are available in three modes - dictionary, pydantic model and as an Excel file.

Dictionaries are created like so:

```python
timeseries_dict = me.outline_metadata('timeseries', mode='dict')
timeseries_dict
```

and updated in the usual way:

```python
timeseries_dict['metadata_information']['producers'][0]['name'] = "example_producer"
timeseries_dict
```

Pydantic is a nice python library for defining and validating data schemas. An outline for the timeseries schema can be created like so:

```python
timeseries_pydantic = me.outline_metadata('timeseries', 'pydantic')
timeseries_pydantic
```

It can be updated using dot notation, for example:

```python
timeseries_pydantic.metadata_information.producers[0].name = "example_producer"
timeseries_pydantic
```

The metadata editor also contains a nice helper function for printing metadata dictionaries and pydantic objects:

```python
me.print_metadata(timeseries_pydantic)
```

Finally, a nicely formatted Excel file can be created into which the metadata can be written:

```python
me.outline_metadata('timeseries', 'excel')
```

## Retreiving existing metadata

Likewise, existing projects can be downloaded as either dictionaries, pydantic models or as excel spreadsheets.

```python
demo_dict = me.get_project_metadata_by_id(timeseries_id, mode='dict')
demo_dict
```

```python
demo_pydantic = me.get_project_metadata_by_id(timeseries_id, mode='pydantic')
demo_pydantic
```

```python
me.get_project_metadata_by_id(timeseries_id, mode='excel', filename=demo_name+'.xlsx')
```

### Updating an existing project

You can create and update projects with metadata as either dictionaries, pydantic models or in Excel files. 

```python
me.update_project_log_by_id(timeseries_id, demo_dict)
```

```python
me.update_project_log_by_id(timeseries_id, demo_pydantic)
```

```python
me.update_project_log_by_id(timeseries_id, demo_name+'.xlsx')
```

# Collections

We can list and create collections, we can add and remove projects from collections and we can apply templates to collections.

```python
me.list_collections().head()
```

```python
collection_id = me.create_collection(demo_name+"_collection", description = "An example collection for demonstration")
```

```python
me.get_collection_by_id(collection_id)
```

## Populating collections

```python
me.add_projects_to_collection(collection=collection_id, id_format='id', projects=[timeseries_id])
```

```python
me.list_projects_in_collection(collection=collection_id, limit=5)
```

```python
me.remove_projects_from_collection(collection=collection_id, id_format='id', projects=[timeseries_id])
```

## Deleting Collections

```python
me.delete_collection_by_id(collection_id)
```

## Deleting Projects

```python
me.delete_project_by_id(timeseries_id)
```

## Explanatory Error Messages

### When the URL is not secure

```python
your_api_key = "bad key"
api_url = "http://example.org/index.php/api/"
me = MetadataEditor(api_url=api_url, api_key=your_api_key)
```

### When the api key is invalid

```python
api_key = "bad_key"
me_bad = MetadataEditor(api_url="https://metadataeditorqa.worldbank.org/index.php/api",
                        api_key=api_key)
me_bad.list_projects(limit=5)
```

### When required data is missing or the type is incorred

```python
series_description = {
                        "idno": 17,
                        "doi": "demoV1",
                     }

me.create_and_log_timeseries(idno="GB20240617_demo3",
                     series_description=series_description)
```

### When the id is incorrect

```python
series_description= {
                "idno": "GB20240617_demo1",
                "doi": "V2",
                "name": "Version 2",
                "display_name": "Version 2",
                "authoring_entity": [{"name": "example agency"}]
                }

me.update_timeseries_by_id(1,
                           series_description=series_description)
```
